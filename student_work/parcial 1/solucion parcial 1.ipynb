```python
def suma_cola(Q):
    suma = 0
    for i in range(len(Q)):
         a = Q.dequeue()
         suma += a
         Q.enqueue(a)
    return suma
```

```python
def intercalar(Q1,Q2):
    nueva_cola = ArrayQueue()
    for i in range(len(Q1)):
        a = Q1.dequeue()
        nueva_cola.enqueue(a)
        b = Q2.dequeue()
        nueva_cola.enqueue(b)
    return nueva_cola
```

```python
Q = ArrayQueue()
Q1 = ArrayQueue()

Q.enqueue(1)
Q.enqueue(2)
Q.enqueue(3)

Q1.enqueue("A")
Q1.enqueue("B")
Q1.enqueue("C")

resultado = intercalar(Q, Q1)

print(resultado._data)
print(Q.is_empty())
print(Q1.is_empty())
```

```python
def contiene (S,Q,x):
    z = False
    for i in range(len(S)):
        a = S.pop()
        if a == x:
            z = True
        Q.enqueue(a)
    for i in range (len(Q)):
        b = Q.dequeue()
        S.push(b)
    return z
```

```python
def contar_apariciones(lista,x):
    if len(lista) == 0:
        return 0
    if lista[0] == x:
        return 1 + contar_apariciones(lista[1:],x)
    else:
        return contar_apariciones(lista[1:],x)
```
