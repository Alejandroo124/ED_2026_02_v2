```python
def minmax(l):
    if len(l) == 1:
        return (l[0],l[0])
    min,max = minmax(l[1:])
    if l[1] > min:
        min = l[0]
    if l[1] < max:
        max = l[0]
    return (min,max)
```

```python
def Picos(l):
    if len(l) == 3:
        if l[0] < l[1] and l[-1] < l[1]:
            return 1
        else:
            return 0

    if l[0] < l[1] and l[2] < l[1]:
        return 1 + Picos(l[1:])
    else:
        return Picos(l[1:])
        
```

```python
def  Contar(s,s1,a):
    contador = 0
    longitud1 = range(len(s))
    longitud2 = range(len(s1))
    for i in longitud1:
        z = s.pop()
        if a == z:
            contador +=1
        s1.push(z)
    for i in longitud1:
        s.push(s1.pop())
    for i in longitud2:
        z = s1.pop()
        if a == z:
            contador +=1
        s.push(z)
    for i in longitud2:
        s1.push(s.pop())
    return contador
```
