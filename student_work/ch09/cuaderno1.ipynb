| Método            | ¿Qué recibe?   | ¿Qué hace?                                                   | Ejemplo              |
| ----------------- | -------------- | ------------------------------------------------------------ | -------------------- |
| `is_empty()`      | Nada           | Indica si la cola está vacía. Devuelve `True` o `False`.     | `pq.is_empty()`      |
| `__len__()`       | Nada           | Devuelve la cantidad de elementos en la cola.                | `len(pq)`            |
| `add(key, value)` | `key`, `value` | Agrega un elemento con una prioridad (`key`).                | `pq.add(1, "Tarea")` |
| `min()`           | Nada           | Consulta el elemento con **menor prioridad** sin eliminarlo. | `pq.min()`           |
| `remove_min()`    | Nada           | Elimina y devuelve el elemento con **menor prioridad**.      | `pq.remove_min()`    |


```python
class StackConHeapPriorityQueue:

    def __init__(self):
        self.datos = HeapPriorityQueue()
        self.prioridad = 0

    def is_empty(self):
        return len(self.datos) == 0

    def __len__(self):
        return len(self.datos)

    def push (self,a):
        self.datos.add(self.prioridad,a)
        self.prioridad -= 1

    def top(self):
        return self.datos.min()[1]
    
    def pop (self):
        if len(self.datos) == 0:
            raise TypeError

        return self.datos.remove_min()[1]
```

```python
s = StackConHeapPriorityQueue()

s.push("A")
s.push("B")
s.push("C")

print(s.pop())
print(s.pop())
print(s.pop())
```
